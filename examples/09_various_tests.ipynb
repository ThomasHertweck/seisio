Obviously, this file is broken (as the name implies). It identifies itself as SEG-Y rev. 0 but seems to have extended textual records (which is not allowed), and __seisio__ has also detected that the length of the file is wrong which is correct as "the file ending is cut out" (quote from __segyio__). __seisio__ tries to read as much as it can and tries to recover as best as possible but clearly, such SEG-Y files should not exist in the first place.

The files "decrement.sgy" and "decrement-lsb.sgy" make no sense as all binary and trace header values are maximized, i.e., they are in a valid range but their actual meaning in SEG-Y context is completely lost and makes no sense any more. __segyio__ opens such a file although the binary header values do not match the actual file size and data but __seisio__ checks, for instance, the number of extended header records, the number of samples, the number of traces in the file as stored in the binary header, etc. and comes up with a weird interpretation of the file and its content. It can still log the binary header or read the first trace, though.

```python
file = datadir + "/" + "decrement.sgy"

with segyio.open(file, ignore_geometry=True) as f:
    log.info("SEGYIO: %s", f.bin)
    
sio = seisio.input(file, ntxtrail=0, nthuser=0, ntxtrec=0)
_ = sio.log_binhead()
trace = sio.read_traces(0)
thstat = sio.thstat(trace)
_ = sio.log_thstat(thstat)
```

__seisio__ in general does not interpret any of the additional header or trailer records. They are simply returned to the user, either decoded or undecoded. 

Conclusion: __seisio__ seems to be able to read most test files without major issues. If headers aren't set correctly, sometimes the user has to supply additional parameters like the number of user-defined trace headers etc. __seisio__ can even read many files that do not conform to the SEG-Y standard (many of the test files actually don't and have, for instance, no fixed-trace-length flag set). If files are severely broken, or binary header values are absolute nonsense, then obviously __seisio__ will struggle. I guess this will hold for any software package. __seisio__, as it stands, won't read 3-byte integer data because it relies heavily on Numpy and there is no corresponding data type available (segyread from the Seismic Unix software package does not seem to read these data formats either) - a workaround would be available, though, if required. As such data formats are rare in practice, currently there is no plan to extend __seisio__ to support them. Get in touch if you have old data based on such formats.
